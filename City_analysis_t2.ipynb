```python
import pandas as pd
df=pd.read_csv("Dataset .csv")

city_counts=df['City'].value_counts()
print("City with highest no.of restaurants:")
print(city_counts.idxmax())
print("\nNumber of restaurants:")
print(city_counts.max())


avg_rating=df.groupby('City')['Aggregate rating'].mean()
print("\nAverage rating for each city:")
print(avg_rating)

highest_avg_city=avg_rating.idxmax()
highest_avg_rating=avg_rating.max()
print("\nCity with highest average rating:")
print(highest_avg_city)
print("\nHighest average rating:")
print(round(highest_avg_rating,2))
```

```python
import matplotlib.pyplot as plt
top_10=avg_rating.head(10)
plt.figure(figsize=(10,5))
top_10.plot(kind='bar')

plt.title("Top 10 cities by avg rating")
plt.xlabel("City")
plt.ylabel("Avr Rating")
plt.show()
plt.savefig("city_analysis.png")
```
