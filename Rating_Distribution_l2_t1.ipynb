```python
import pandas as pd
df=pd.read_csv("Dataset .csv")
df[['Aggregate rating','Votes']].head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.hist(df['Aggregate rating'], bins=10)
plt.title("Distribution of Aggregate Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.savefig("Rating_distribution.png")
plt.show()
```

```python
rating_ranges=pd.cut(df['Aggregate rating'], bins=5)
rating_counts=rating_ranges.value_counts()
print(rating_counts)
print("\nMost common rating range:")
print(rating_counts.idxmax())

avg_votes=df['Votes'].mean()
print("Average Votes Received:")
print(round(avg_votes,2))
```
