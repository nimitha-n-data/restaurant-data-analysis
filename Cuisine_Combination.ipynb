```python
import pandas as pd
df=pd.read_csv("Dataset .csv")
cuisine_counts=df['Cuisines'].value_counts()
print("Top 10 cuisine combination:")
print(cuisine_counts.head(10))
avg_rating=df.groupby('Cuisines')['Aggregate rating'].mean()
print("Average Rating by Cuisine Combination:")
print(avg_rating.sort_values(ascending=False).head(10))
```

```python
import matplotlib.pyplot as plt
top10 = cuisine_counts.head(10)
plt.figure(figsize=(12,6))
top10.plot(kind='bar')
plt.title("Top 10 Cuisine Combinations")
plt.xlabel("Cuisine Combinations")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45, ha='right')
plt.savefig("CUisine_combination.png")
plt.show()
```
