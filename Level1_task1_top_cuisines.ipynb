```python
import pandas as pd
from collections import Counter

df = pd.read_csv("Dataset .csv")
cuisine_counter=Counter()

for cuisines in df['Cuisines'].dropna():
  for cuisine in cuisines.split(','):
    cuisine_counter[cuisine.strip()]+=1

top_3=cuisine_counter.most_common(3)
total_restaurants=len(df)

for cuisine, count in top_3:
  percentage=(count/total_restaurants)*100
  print(cuisine,count,round(percentage,2))
```

```python
import matplotlib.pyplot as plt
cuisines = [item[0] for item in top_3]
counts = [item[1] for item in top_3]
plt.bar(cuisines,counts)
plt.title("Top 3 Most Common Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Number of restaurents")
plt.savefig("top_cuisines.png")
plt.show()
```
