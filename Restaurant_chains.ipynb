```python
import pandas as pd
df=pd.read_csv("Dataset .csv")
chain_counts=df['Restaurant Name'].value_counts()
chains = chain_counts[chain_counts > 1]
print("Restaurant Chains:")
print(chains.head(10))
chain_ratings = df.groupby('Restaurant Name')['Aggregate rating'].mean()
print("Average Ratings of Restaurant Chains:")
print(chain_ratings.loc[chains.index].sort_values(ascending=False).head(10))
chain_votes=df.groupby('Restaurant Name')['Votes'].sum()
print("Most Popular Restaurant chains:")
print(chain_votes.loc[chains.index].sort_values(ascending=False).head(10))
```

```python
import matplotlib.pyplot as plt
top_chains = chains.head(10)
plt.figure(figsize=(10,6))
top_chains.plot(kind='bar')
plt.title("Top 10 Restaurant Chains")
plt.xlabel("Restaurant Chain")
plt.ylabel("Number of branches")
plt.xticks(rotation=45,ha='right')
plt.savefig("restaurant_chains.png")
plt.show()
```
